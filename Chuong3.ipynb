Trang này minh họa việc thực thi code Python để tạo biểu đồ, nhưng nhờ cấu hình tag `remove-input`, độc giả sẽ **chỉ nhìn thấy hình vẽ** mà không bị rối mắt bởi đoạn code bên dưới.

```python
import matplotlib.pyplot as plt
import numpy as np

# Tạo dữ liệu mẫu
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Vẽ đồ thị
plt.figure(figsize=(8, 4))
plt.plot(x, y, label='Sóng Sine', color='#0f766e', linewidth=2)
plt.title('Biểu đồ Dao động Sine', fontsize=14)
plt.xlabel('Trục X (Thời gian)')
plt.ylabel('Trục Y (Biên độ)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
